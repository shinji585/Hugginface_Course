Key words: 

LLMs (Large Language Models): modelos de deep learning entrenados con grandes volúmenes de texto que operan sobre tokens (números) para predecir el siguiente token y generar texto.

NLP: rama de la inteligencia artificial que estudia cómo las computadoras procesan, analizan y generan lenguaje humano mediante métodos matemáticos y estadísticos.

Tokens: unidades de texto convertidas en números por un tokenizer; pueden ser palabras, subpalabras o caracteres.

Transformer: arquitectura de deep learning basada en mecanismos de atención. Muchos LLMs modernos están basados en esta arquitectura.

Transformers: librería de Hugging Face que permite usar modelos Transformer preentrenados.

Datasets: librería para cargar y manejar grandes conjuntos de datos de forma eficiente.

```python
# A continuacion se estara utilizando la funcion pipeline que nos permite cargar una task y ella misma se encarga de elegir el modelo el numero de tokens adecuado y las reglas de entrada y salidad para dicho modelo 
from transformers import pipeline

# definimos la task 
task: str =  "sentiment-analysis"
# vamos a seguir el mismo ejemplo que el curso entonces estaremos creando un analizador de sentimiento 
classifier = pipeline(task=task)

# despues de cargar el modelo nosotros podemos llamar y pasarle texto y este analiza los sentimiento del texto 
result = classifier("Today is not a good day,I hate everything and I want to stop with this")
print(result)
```
