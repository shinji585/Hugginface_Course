Key words:

LLMs (Large Language Models): modelos de deep learning entrenados con grandes volúmenes de texto que operan sobre tokens (números) para predecir el siguiente token y generar texto.

NLP: rama de la inteligencia artificial que estudia cómo las computadoras procesan, analizan y generan lenguaje humano mediante métodos matemáticos y estadísticos.

Tokens: unidades de texto convertidas en números por un tokenizer; pueden ser palabras, subpalabras o caracteres.

Transformer: arquitectura de deep learning basada en mecanismos de atención. Muchos LLMs modernos están basados en esta arquitectura.

Transformers: librería de Hugging Face que permite usar modelos Transformer preentrenados.

Datasets: librería para cargar y manejar grandes conjuntos de datos de forma eficiente.

```python
# A continuacion se estara utilizando la funcion pipeline que nos permite cargar una task y ella misma se encarga de elegir el modelo el numero de tokens adecuado y las reglas de entrada y salidad para dicho modelo
from transformers import pipeline

# definimos la task
task: str =  "sentiment-analysis"
# vamos a seguir el mismo ejemplo que el curso entonces estaremos creando un analizador de sentimiento
classifier = pipeline(task=task)

# despues de cargar el modelo nosotros podemos llamar y pasarle texto y este analiza los sentimiento del texto
result = classifier("Today is not a good day,I hate everything and I want to stop with this")
print(result)
```

```python
# podemos pasarle multiples sentencias a el modelo
sentences: list[str] = ["Today is a greate day and I feel very good","Today is not the best day but I'm going to put all my effor again"]

# le pasamos la sentencias y este nos devuelve una lista que adentro tiene un diccionario (simulando un json) con el label que es la etiqueta y el score que es cuan seguro esta el modelo
result = classifier(sentences)
print(result) # retorna una sola lista con los datos
```

```python
# ahora veremos un tipo de clasificacion que es zero-shot-classification que permite que un modelo eliga y clasifique texto dependiendo de un contexto pasado a este
from transformers import pipeline

# creamos el objecto
zero = pipeline(task="zero-shot-classification")
etiquetas: list[str] = ["education","health","lifestyle","social media"]
# le pasamos un contexto
result = result = zero(
    "Make exercise is the best for your life if you don't do that maybe you could have problems in the future with your health",
    candidate_labels=etiquetas
)

for k,v in result.items():
  print(f"\n{k}: {v}\n")
```

```python
# otra tarea la cual podemos realizar con estos modelos es la generacion de texto aleatorio esto es lo que pasa cuando nosotros escribimos en nuestros telefonos
# el modelo mira lo que ya escribimos, calcula que palabra es mas probable que siga y la sugiere
from transformers import pipeline
import pprint

# creamos el objecto
chat = pipeline(task="text-generation")

result = chat(
    "In this section, I'm trying to use text generator"
)

# mostramos el resultado
pprint.pprint(result)
```
